## Activity 1.01 Handling Lists

Task: generate a list of 100 random numbers, build a second list holding only the numbers divisible by three, compare the lengths of the two lists, then repeat the experiment to see how stable that ratio is.

```python
# Activity 1.01 - Step 1: make a list of 100 random numbers
import random

random.seed(1)  # this makes the random numbers come out the same every time I run it

random_list = []
for i in range(100):
    random_list.append(random.randint(0, 100))

print("How many numbers I made:", len(random_list))
print("The first 20 of them:")
print(random_list[:20])
```

```python
# Step 2: make a second list with only the numbers that divide evenly by 3
divisible_by_three = []
for number in random_list:
    if number % 3 == 0:
        divisible_by_three.append(number)

print("Numbers divisible by 3:")
print(divisible_by_three)
print()

# Step 3: compare how long the two lists are
print("Length of the first list:", len(random_list))
print("Length of the second list:", len(divisible_by_three))
print("First length divided by second length:", round(len(random_list) / len(divisible_by_three), 2))
```

```python
# Repeat looking for any changes
for trial in range(5):
    trial_list = []
    for i in range(100):
        trial_list.append(random.randint(0, 100))

    trial_multiples = []
    for number in trial_list:
        if number % 3 == 0:
            trial_multiples.append(number)

    print("Trial", trial + 1, "found", len(trial_multiples), "multiples of 3, ratio:", round(100 / len(trial_multiples), 2))

print()
```

## Activity 1.02 Analyzing a Multiline String and Generating the Unique Word Count


```python
# Activity 1.02 - Step 1: start with a multiline string
# Triple quotes let me keep the line breaks exactly how I typed them
multiline_text = """It was the best of times, it was the worst of times, it was
the age of wisdom, it was the age of foolishness, it was the epoch of belief,
it was the epoch of incredulity, it was the season of Light, it was the season
of Darkness, it was the spring of hope, it was the winter of despair, we had
everything before us, we had nothing before us, we were all going direct to
Heaven, we were all going direct the other way - in short, the period was so
far like the present period, that some of its noisiest authorities insisted on
its being received, for good or for evil, in the superlative degree of
comparison only."""

print("How many characters:", len(multiline_text))
print("How many line breaks:", multiline_text.count("\n"))
print()
print(multiline_text)
```

```python
# Step 2: take out the line breaks
# I swap each line break for a space so words on two lines do not get stuck together
flattened = multiline_text.replace("\n", " ")

print("Line breaks left:", flattened.count("\n"))
print()
print(flattened)
```

```python
# Step 3: split the text into sentences
# A period ends a sentence, so I split the text everywhere there is a period
sentences = flattened.split(".")

print("Pieces after splitting:", len(sentences))
for sentence in sentences:
    print(sentence)
```

```python
# split() with nothing inside the parentheses splits on spaces
all_words = []
for sentence in sentences:
    for word in sentence.split():
        all_words.append(word)

# Looking at the list size we can find the amount of words
print("Total words:", len(all_words))
print(all_words)
```

```python
# Clean up the data
import string

cleaned_words = []
for word in all_words:
    clean = ""
    for letter in word:
        # string.punctuation is a list of marks like , . ! ? so I skip those letters
        if letter not in string.punctuation:
            clean = clean + letter
    clean = clean.lower()  # make the clean value lowercase
    if clean != "":        # this drops anything that was only punctuation
        cleaned_words.append(clean)

print("Words before cleaning:", len(all_words))
print("Words after cleaning:", len(cleaned_words))
print("The first 15 cleaned words:")
print(cleaned_words[:15])
```

```python
# Count the unique words
# The word is the key and the number of times I saw it is the value
word_counts = {}
# We pretty much just check if it exists in the dictionary. If it does we add 1 otherwise set it to 1.
for word in cleaned_words:
    if word in word_counts:
        word_counts[word] = word_counts[word] + 1
    else:
        word_counts[word] = 1

print("Total words:", len(cleaned_words))
print("Unique words:", len(word_counts))
print()
print("Each word and how many times it shows up:")
for word in word_counts:
    print(word, word_counts[word])
```

## Activity 2.01 Permutation, Iterator, Lambda, List

```python
# Imports for permutations library and initializations.
from itertools import permutations, dropwhile
permutations
dropwhile

# Grab 3 ( I think this is a memory location)
permutations(range(3))
```

```python
# Loop through and list them all
for number_tuple in permutations(range(3)):
    print(number_tuple)
    assert isinstance(number_tuple, tuple)  
```

```python
# This cleans it up by dropping the 0's and making it a list
for number_tuple in permutations(range(3)):
    print(list(dropwhile(lambda x: x <= 0, number_tuple))) 
```

```python
import math


# This will concatenate the digits and convert that string into a number
def convert_to_number(number_stack):
    final_number = 0
    for i in range(0, len(number_stack)):
        final_number += (number_stack.pop() * (math.pow(10, i)))
    return final_number

# This will loop through each permutation and convert it.
for number_tuple in permutations(range(3)):
    number_stack = list(dropwhile(lambda x: x <= 0, number_tuple))
    print(convert_to_number(number_stack))  
```

## Activity 2.02 Designing Your Own CSV Parser

```python
from itertools import zip_longest

# This will read in the csv compressing the lines to be better in memory.
def return_dict_from_csv_line(header, line):
    # Zip them
    zipped_line = zip_longest(header, line, fillvalue=None)
    # Use dict comprehension to generate the final dict
    ret_dict = {kv[0]: kv[1] for kv in zipped_line}
    return ret_dict  

# Open and split up the values by the comma
with open("./data/sales_record.csv", "r") as fd:
    first_line = fd.readline()
    header = first_line.replace("\n", "").split(",")
    for i, line in enumerate(fd):
        # Here we loop over the first 10 lines in order to not to make the output too big
        # This dataset is very large so this make sense
        # Also this will remove the new line characters. With what we are doing including these could mess up the output
        line = line.replace("\n", "").split(",")
        d = return_dict_from_csv_line(header, line)
        print(d)
        if i > 10:
            break
```
