```python
# Imports
import sqlite3
# Connect to the database
conn = sqlite3.connect("data/petsdb")
```

```python
# This iterates through the query getting the count of people in each age
for ppl, age in conn.execute("SELECT count(*), age FROM persons GROUP BY age"):
    print("We have {} people aged {}".format(ppl, age))
```

```python
# This query adds to the previous one by checking the age with the most people and returning that age and the number of people
for ppl, age in conn.execute("SELECT * FROM (SELECT count(*) as count, age FROM persons GROUP BY age) ORDER BY count DESC LIMIT 1"):
    print("The age with the most people is {} with {} people".format(age, ppl))
```

```python
# This gets the count of people who have null for their last_name in the database
no_last_name_count = conn.execute("SELECT count(*) FROM persons WHERE last_name is null").fetchone()[0]
print("{} people have no last name".format(no_last_name_count))
```

```python
multiple_pet_count = conn.execute("SELECT count(*) FROM persons WHERE id IN (SELECT owner_id FROM pets GROUP BY owner_id HAVING count(*) > 1)").fetchone()[0]
print("{} people have more than one pet".format(multiple_pet_count))
```

```python
treatment_count = conn.execute("SELECT count(*) from pets where treatment_done != 0").fetchone()[0]
print("{} pets have received treatment".format(treatment_count))
```

```python
treatment_type_count = conn.execute("SELECT count(*) from pets where treatment_done != 0 and pet_type is not null").fetchone()[0]
print("{} pets have received treatment and their type is specified".format(treatment_type_count))
```

```python
east_port_count = conn.execute("select count(*) from pets where owner_id in (select id from persons where city = 'east port')").fetchone()[0]
print("{} pets have owners who live in East Port".format(east_port_count))
```

```python
east_port_treatment_count = conn.execute("select count(*) from pets where owner_id in (select id from persons where city = 'east port') and treatment_done != 0").fetchone()[0]
print("{} pets have owners who live in East Port and have received treatment".format(east_port_treatment_count))
```
