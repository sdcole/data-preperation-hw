## Setup

```python
# Imports
import numpy
import pandas as pd
```

```python
# Load data

# I had to mess with the encoding to get this to work since for some reason it was not in utf8.
candy_data = pd.read_csv("data/candyhierarchy2017.csv", encoding="mac_roman")

art_data = pd.read_csv("data/MetObjects.csv")
```

## Chapter 7

```python
# Get rid of missing data
# The columns that start with "Q6 |" are the columns that contain the ratings for each candy. We want to drop any rows that have all missing values in these columns.
rating_cols = [c for c in candy_data.columns if c.startswith("Q6 |")]
candy_data = candy_data.dropna(how="all", subset=rating_cols)

candy_data.head(10)
```

```python
# Check and remove duplicates
# None are in this dataset
print(candy_data["Internal ID"].duplicated().sum())
```

## Chapter 8

### Reshape and create a hierarchical index

```python
# Reshape the candy data from wide to long
# Right now each candy is its own "Q6 |" column. Melting turns it into one row per (respondent, candy) with the rating in a single column.
candy_long = candy_data.melt(
    id_vars=["Internal ID", "Q1: GOING OUT?", "Q2: GENDER", "Q3: AGE", "Q4: COUNTRY"],
    value_vars=rating_cols,
    var_name="Candy",
    value_name="Rating",
)

# Clean up the candy names and drop candies the respondent skipped
candy_long["Candy"] = candy_long["Candy"].str.removeprefix("Q6 | ")
candy_long = candy_long.dropna(subset=["Rating"])

# Hierarchical index: respondent on the outer level, candy on the inner level
candy_long = candy_long.set_index(["Internal ID", "Candy"]).sort_index()

print(candy_data.shape, "->", candy_long.shape)
candy_long.head(10)
```

```python
# Use the hierarchical index to pull out every rating for a single candy
candy_long.xs("Kit Kat", level="Candy")["Rating"].value_counts()
```

## Chapter 10

```python
# Group the data by the index levels
candy_long.groupby(level="Candy")["Rating"].count().sort_values(ascending=False).head(10)
```

```python
# Cross tab the data
pd.crosstab(candy_long.index.get_level_values("Candy"), candy_long["Rating"], rownames=["Candy"])
```

## Chapter 11

```python
# Convert between string and datetime
# "Metadata Date" is stored as text like "3/6/2025 4:54:30 AM", so parse it into a real datetime column
art_data["Metadata Date"] = pd.to_datetime(art_data["Metadata Date"], format="%m/%d/%Y %I:%M:%S %p")
print(art_data["Metadata Date"].dtype)

# And back to a string in a different format
art_data["Metadata Date"].dt.strftime("%Y-%m-%d").head(10)
```

```python
# Generate a date range
# One timestamp for the start of every month between the oldest and newest metadata update
start = art_data["Metadata Date"].min().replace(day=1).normalize()
end = art_data["Metadata Date"].max()
months = pd.date_range(start=start, end=end, freq="MS")

# Count how many records were updated each month, lined up against the full range so months with no updates show 0
updates_per_month = (
    art_data.groupby(pd.Grouper(key="Metadata Date", freq="MS"))
    .size()
    .reindex(months, fill_value=0)
)
updates_per_month
```
